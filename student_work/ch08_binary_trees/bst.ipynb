```python
def find_min(T:LinkedBinaryTree):
    '''Retorna el contenedor donde se encuentra el minimo'''
    current = T.root()
    while T.left(current) is not None:
        current = T.left(current)
    return current
    
```

```python
find_min(L).element()
```

```python
def find_max(T:LinkedBinaryTree):
    '''Retorna el contenedor donde se encuentra el minimo'''
    current = T.root()
    while T.right(current) is not None:
        current = T.right(current)
    return current
```

```python
find_max(L).element()
```

```python
def es_ordenado(T:LinkedBinaryTree):
    I = T.inorder()
    a = next(I)
    for i in range(len(T) - 2):
        b = next(I)
        if b.element() < a.element():
            return False 
        a = next(I)
        a, b = b, a

    return True 
```

```python
def es_ordenado1(T:LinkedBinaryTree):
    I = T.inorder()
    I1 = T.inorder()
    next(I1)
    for i in range(len(T) - 1):
        if next(I).element() > next(I).element():
            return False
    return True
```
