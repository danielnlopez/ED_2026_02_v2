## Capítulo 7: Lista enlazadas

```python
class SinglyLinkedList:
    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    class _Node:
        __slots__ = '_element', '_next'
        def __init__(self, element, nxt):
            self._element = element
            self._next    = nxt

    def __len__(self):
        return self._size
    
    def is_empty(self):
        return self._size == 0
    
    def first(self):
        return self._head._element

    def insert_first(self, e):
        self._head = self._Node(e, self._head)
        if self._size == 0:
            self._tail = self._head
        self._size += 1

    def insert_last(self, e):
        node = self._Node(e, None)
        if self.is_empty():
            self._head = node
        else:
            self._tail._next = node
        self._tail = node
        self._size += 1

    def remove_first(self):
        val = self._head._element
        self._head = self._head._next
        self._size -= 1
        if self._size == 0:
            self._tail = None
        return val

S = SinglyLinkedList()
for x in ['A', 'B', 'C', 'D']:
    S.insert_last(x)
print(len(S), S.first())    # 4  A
S.remove_first()
print(len(S), S.first())    # 3  B
```

```python
def build(items):
    """Construye una SinglyLinkedList con los elementos dados."""
    L = SinglyLinkedList()
    for x in items:
        L.insert_last(x)
    return L

def to_list(L):
    """Recorre L desde la cabeza y retorna sus elementos en una lista de Python."""
    out, node = [], L._head
    while node is not None:
        out.append(node._element)
        node = node._next
    return out
```

**R-7.1** Give an algorithm for finding the second-to-last node in a singly linked list in which the last node is indicated by a `next` reference of `None`.

```python
def second_to_last(L):
    """Retorna el penúltimo nodo (_Node) de L, o None si L tiene menos de 2 elementos."""

    if len(L) < 2:
        return None
    
    if L._head is None or L._head._next is None:
            raise ValueError("La lista debe tener al menos 2 elementos") 
    
    current = L._head

    for i in range(len(L) - 2):
        current = current._next

    return current
```

```python
def concatenate(L:SinglyLinkedList, M: SinglyLinkedList):
    """Retorna una SinglyLinkedList con todos los nodos de L seguidos de todos los nodos de M.
    Debe reutilizar los nodos existentes (no crear nodos nuevos)."""

    L._tail._next = M._head
    L._tail = M._tail
    L._size += M._size

    return L
   
```

```python
from goodrich.exceptions import Empty

class LinkedStack:
    """Pila implementada con la SinglyLinkedList del cuaderno (patrón adaptador).
    La lista se guarda en self._list y el tope de la pila es la cabeza de la lista."""

    def __init__(self):
        self._list = SinglyLinkedList()

    def __len__(self):
        return len(self._list)

    def is_empty(self):
        return self._list.is_empty()

    def push(self, e):
        self._list.insert_first(e)

    def top(self):
        if self.is_empty():
            raise Empty("La pila está vacía")
        return self._list.first()

    def pop(self):
        if self.is_empty():
            raise Empty("La pila está vacía")
        return self._list.remove_first()
```

**R-9.2** Suppose you label each position $p$ of a binary tree $T$ with a key equal to its preorder rank. Under what circumstances is $T$ a heap?

```python
from goodrich.ch06.array_queue import ArrayQueue

def label_preorder(T):
    """Reemplaza el elemento de cada posición por su rango en preorden (0, 1, 2, ...)."""
    for rank, p in enumerate(T.preorder()):
        T._replace(p, rank)

def has_heap_order(T):
    return all(p.element() <= c.element() for p in T.positions() for c in T.children(p))

def is_complete(T):
    """True si T es un árbol binario completo (BFS: después de un hueco no puede haber nodos)."""
    if T.is_empty():
        return True
    queue, gap = ArrayQueue(), False
    queue.enqueue(T.root())
    while not queue.is_empty():
        p = queue.dequeue()
        for c in (T.left(p), T.right(p)):
            if c is None:
                gap = True
            elif gap:
                return False
            else:
                queue.enqueue(c)
    return True

def preorder_labeled_is_heap(T):
    label_preorder(T)
    return has_heap_order(T) and is_complete(T)

# Respuesta: en preorden un padre siempre se visita antes que sus descendientes, así que
# el orden de heap se cumple SIEMPRE. Por lo tanto T es un heap si y solo si T es completo.
```

**C-9.26** Show how to implement the stack ADT using only a priority queue and one additional integer instance variable.

```python
from goodrich.exceptions import Empty
from goodrich.ch09.heap_priority_queue import HeapPriorityQueue

class PQStack:
    """Pila implementada con una cola de prioridad y un contador entero."""

    def __init__(self):
        self._data = HeapPriorityQueue()
        self._count = 0

    def __len__(self):
        return len(self._data)

    def is_empty(self):
        return len(self) == 0

    def push(self, e):
        self._count += 1
        self._data.add(-self._count, e)

    def top(self):
        if self.is_empty():
            raise Empty("La pila está vacía")
        return self._data.min()[1]

    def pop(self):
        if self.is_empty():
            raise Empty("La pila está vacía")
        return self._data.remove_min()[1]
```

```python
class PQQueue:
    """Cola FIFO implementada con una cola de prioridad y un contador entero."""

    def __init__(self):
        self._data = HeapPriorityQueue()
        self._count = 0

    def __len__(self):
        return len(self._data)

    def is_empty(self):
        return len(self) == 0

    def enqueue(self, e):
        self._count += 1
        self._data.add(self._count, e)

    def first(self):
        """Lanza Empty si la cola está vacía."""
        if self.is_empty():
             raise Empty("La cola está vacía")
        return self._data.min()[1]

    def dequeue(self):
        """Lanza Empty si la cola está vacía."""
        if self.is_empty():
            raise Empty("La cola está vacía")
        return self._data.remove_min()[1]
```

```python
# Pruebas C-9.27
Q = PQQueue()
assert len(Q) == 0 and Q.is_empty()
for metodo in (Q.first, Q.dequeue):
    try:
        metodo()
        assert False, f'{metodo.__name__} en cola vacía debe lanzar Empty'
    except Empty:
        pass

Q.enqueue('a'); Q.enqueue('b'); Q.enqueue('c')
assert len(Q) == 3 and Q.first() == 'a'
assert Q.dequeue() == 'a' and Q.dequeue() == 'b'
Q.enqueue('d')                                  # se intercalan inserciones y extracciones
assert [Q.dequeue(), Q.dequeue()] == ['c', 'd']
assert Q.is_empty()

for i in range(1000):
    Q.enqueue(i)
assert [Q.dequeue() for _ in range(1000)] == list(range(1000))
print('C-9.27: todas las pruebas pasaron')
```

### 1. Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo

```python
def contar_nodos_un_hijo(T: LinkedBinaryTree):
    contador = 0

    for p in T.positions():
        # Verificar si tiene hijo izquierdo y derecho
        izquierdo = T.left(p)
        derecho = T.right(p)

        if (izquierdo is None and derecho is not None) or (izquierdo is not None and derecho is None):
            contador += 1

    return contador
```

### 2. Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo izquierdo

```python
def contar_nodos_hijo_izquierdo(T: LinkedBinaryTree):
    contador = 0

    for p in T.positions():
        if T.left(p) is not None and T.right(p) is None:
            contador += 1

    return contador
```

### 3. Hallar el número de nodos en un linkedbinarytree que solo tienen un hijo derecho

```python
def contar_nodos_hijo_derecho(T: LinkedBinaryTree):
    contador = 0

    for p in T.positions():
        if T.right(p) is not None and T.left(p) is None:
            contador += 1

    return contador
```

### 4. Tomar con cuidado: Saber si dos linkedbinarytree son iguales

```python
def arboles_iguales(T1: LinkedBinaryTree, T2: LinkedBinaryTree):
    if len(T1) != len(T2):
        return False

    def iguales(p1, p2):
        if p1 is None and p2 is None:
            return True

        if p1 is None or p2 is None:
            return False

        if T1.element(p1) != T2.element(p2):
            return False

        return (iguales(T1.left(p1), T2.left(p2)) and
                iguales(T1.right(p1), T2.right(p2)))

    if len(T1) == 0:
        return True

    return iguales(T1.root(), T2.root())
```

### 5. Obtener una lista con todos los nodos que se encuentran a un mismo nivel, leidos de izquierda a derecha en un LBT

```python
def nodos_nivel(T: LinkedBinaryTree, p, nivel):
    if p is None:
        return []

    if nivel == 0:
        return [T.element(p)]

    return (nodos_nivel(T, T.left(p), nivel - 1) +
            nodos_nivel(T, T.right(p), nivel - 1))
```
